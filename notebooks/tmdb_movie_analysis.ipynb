## Step 0.2 — Parsing JSON String Columns

Some columns in the `movies` and `credits` DataFrames look like Python lists or dicts but are actually **plain strings**.
In this step we use `ast.literal_eval` to convert these strings into real Python objects.

- `movies`: `genres`, `keywords`, `production_companies`, `production_countries`, `spoken_languages`
- `credits`: `cast`, `crew`

For broken / unparseable values we safely assign an **empty list** via `try/except`.
At the end we print the number of errors for each column along with sample values.

```python
def safe_parse(val):
    # Parse string → Python object
    try:
        return ast.literal_eval(val)
    except (ValueError, SyntaxError):
        return []

# ── Parse movies JSON columns ────────────────────
json_cols_movies = [
    'genres', 'keywords', 'production_companies',
    'production_countries', 'spoken_languages'
]

error_counts = {}
for col in json_cols_movies:
    # Count unparseable rows before converting
    errors = movies[col].apply(
        lambda v: not isinstance(safe_parse(v) if isinstance(v, str) else v, list)
    ).sum()
    error_counts[col] = errors
    movies[col] = movies[col].apply(safe_parse)

# ── Parse credits JSON columns ───────────────────
json_cols_credits = ['cast', 'crew']

for col in json_cols_credits:
    errors = credits[col].apply(
        lambda v: not isinstance(safe_parse(v) if isinstance(v, str) else v, list)
    ).sum()
    error_counts[col] = errors
    credits[col] = credits[col].apply(safe_parse)

# ── Verification ─────────────────────────────────
print("── Parse error counts (should all be 0) ──")
for col, cnt in error_counts.items():
    print(f"  {col}: {cnt} errors")

print("\n── Type checks (should all be <class 'list'>) ──")
for col in json_cols_movies:
    print(f"  movies['{col}'][0] → {type(movies[col][0])}")
for col in json_cols_credits:
    print(f"  credits['{col}'][0] → {type(credits[col][0])}")

print("\n── Sample: movies['genres'][0] ──")
print(movies['genres'][0])

print("\n── Sample: movies['keywords'][0][:3] ──")
print(movies['keywords'][0][:3])

print("\n── Sample: credits['cast'][0][:2] ──")
print(credits['cast'][0][:2])

print("\n── Sample: credits['crew'][0][:2] ──")
print(credits['crew'][0][:2])
```

### 📋 Step 0.2 — Findings & Decisions

|---|-------|-------|

| 2 | All columns now `<class 'list'>` | Ready for field extraction in Step 0.3 |
| 3 | `genres` → structure `[{'id': 28, 'name': 'Action'}, ...]` | The `name` field will be extracted → `genres_list` |

## Step 0.3 — Extracting Structured Fields from Parsed Columns

We extract the information needed for analysis and modeling from the parsed JSON columns:

- `genres` → list of genre names → `genres_list`
- `keywords` → list of keyword names → `keywords_list`
- `production_companies` → list of studio names → `studios_list`
- `cast` → names of the top 3 actors by `order` → `actor_1`, `actor_2`, `actor_3`
- `crew` → name of the person with `job == 'Director'` → `director`

We **keep** the original parsed columns as a backup (they will be dropped after the merge).

```python
# ── Extract name lists from movies ───────────────
def extract_names(lst):
    # Get 'name' field from each dict
    return [d['name'] for d in lst if isinstance(d, dict) and 'name' in d]

movies['genres_list']   = movies['genres'].apply(extract_names)
movies['keywords_list'] = movies['keywords'].apply(extract_names)
movies['studios_list']  = movies['production_companies'].apply(extract_names)

# ── Extract top 3 actors from credits ────────────
def get_top_actors(cast_list, n=3):
    # Sort by order, take top n names
    sorted_cast = sorted(cast_list, key=lambda d: d.get('order', 999))
    return [d['name'] for d in sorted_cast[:n] if isinstance(d, dict)]

top_actors = credits['cast'].apply(get_top_actors)

credits['actor_1'] = top_actors.apply(lambda x: x[0] if len(x) > 0 else None)
credits['actor_2'] = top_actors.apply(lambda x: x[1] if len(x) > 1 else None)
credits['actor_3'] = top_actors.apply(lambda x: x[2] if len(x) > 2 else None)

# ── Extract director from crew ────────────────────
def get_director(crew_list):
    # Find first crew member with job = Director
    for member in crew_list:
        if isinstance(member, dict) and member.get('job') == 'Director':
            return member.get('name')
    return None

credits['director'] = credits['crew'].apply(get_director)

# ── Verification ──────────────────────────────────
print("── movies new columns sample ──")
print(movies[['title', 'genres_list', 'keywords_list', 'studios_list']].head(3).to_string())

print("\n── credits new columns sample ──")
print(credits[['title', 'actor_1', 'actor_2', 'actor_3', 'director']].head(5).to_string())

print("\n── Null / empty counts in extracted columns ──")
for col in ['genres_list', 'keywords_list', 'studios_list']:
    empty = (movies[col].apply(len) == 0).sum()
    print(f"  movies['{col}'] empty lists: {empty}")

for col in ['actor_1', 'actor_2', 'actor_3', 'director']:
    print(f"  credits['{col}'] nulls: {credits[col].isnull().sum()}")
```

| 2 | `keywords_list` empty list | **412** movies (8.6%) | High but normal — movies without keywords will be filtered out in the RQ#8 analysis |

## Step 0.4 — Merging the Two DataFrames

We merge the `movies` and `credits` tables into a single DataFrame.

- Merge key: `movies.id` ↔ `credits.movie_id`
- Method: **inner join** — only rows that match in both tables are kept
- After the merge, we check:
  - Row count before / after
  - Are there any duplicate rows?
  - How many rows were lost?
- Duplicate columns coming from the merge are cleaned up: `movie_id` (= `id`) and `title_y` (= `title_x`)

```python
print(f"Before merge — movies: {movies.shape}, credits: {credits.shape}")

# ── Inner join on id / movie_id ───────────────────
df = movies.merge(credits, left_on='id', right_on='movie_id', how='inner')

print(f"After merge  — df: {df.shape}")
print(f"Rows lost    — {len(movies) - len(df)}")

# ── Check duplicate rows (skip list columns) ──────
# List columns are unhashable — exclude them
list_cols = [c for c in df.columns
             if df[c].apply(lambda x: isinstance(x, list)).any()]
scalar_cols = [c for c in df.columns if c not in list_cols]
n_dupes = df[scalar_cols].duplicated().sum()
print(f"\nFull duplicate rows: {n_dupes}")

# ── Check duplicate movie ids ─────────────────────
n_id_dupes = df['id'].duplicated().sum()
print(f"Duplicate movie ids: {n_id_dupes}")

# ── Drop redundant columns ────────────────────────
# movie_id == id  /  title_y == title_x
cols_to_drop = ['movie_id']
if 'title_x' in df.columns and 'title_y' in df.columns:
    cols_to_drop.append('title_y')
    df = df.rename(columns={'title_x': 'title'})

df = df.drop(columns=cols_to_drop)

print(f"\nDropped columns : {cols_to_drop}")
print(f"Final df shape  : {df.shape}")
print(f"\nColumns ({len(df.columns)}):")
print(list(df.columns))
```

## Step 0.5 — Handling Missing Values

We handle missing/invalid values in each column according to its intended use:

- `budget` / `revenue`: 0 → `NaN`; rows where both are NaN are dropped
- `runtime`: 0 → `NaN`; remaining nulls are filled with the median
- `release_date`: the row with a null value is dropped (only 1)
- `tagline`: null → `''`; a `has_tagline` flag column is added (for RQ#10)
- `homepage` / `overview`: null → `''` (won't be used in ML)
- `actor_1/2/3` / `director`: null → `'Unknown'`
- At the end, the null count for every column is printed again

```python
# ── budget & revenue: 0 → NaN ───────────────────
df['budget']  = df['budget'].replace(0, np.nan).astype('float64')
df['revenue'] = df['revenue'].replace(0, np.nan).astype('float64')

# Drop rows where BOTH budget and revenue are NaN
before = len(df)
df = df[~(df['budget'].isna() & df['revenue'].isna())]
print(f"Rows dropped (budget AND revenue NaN): {before - len(df)}")

# ── runtime: 0 → NaN → median impute ─────────────
df['runtime'] = df['runtime'].replace(0, np.nan)
runtime_median = df['runtime'].median()
df['runtime'] = df['runtime'].fillna(runtime_median)
print(f"runtime median used for imputation: {runtime_median}")

# ── release_date: drop null rows ─────────────────
before = len(df)
df = df.dropna(subset=['release_date'])
print(f"Rows dropped (release_date null): {before - len(df)}")

# ── tagline: fill + flag ──────────────────────────
df['has_tagline'] = df['tagline'].notna().astype(int)
df['tagline'] = df['tagline'].fillna('')

# ── homepage & overview: fill with '' ────────────
df['homepage'] = df['homepage'].fillna('')
df['overview'] = df['overview'].fillna('')

# ── actor_1/2/3, director: fill with 'Unknown' ───
for col in ['actor_1', 'actor_2', 'actor_3', 'director']:
    df[col] = df[col].fillna('Unknown')

# ── Final null summary ────────────────────────────
print(f"\nFinal df shape: {df.shape}")
null_final = df.isnull().sum()
null_final = null_final[null_final > 0]
if null_final.empty:
    print("\nNo nulls remaining — all columns clean.")
else:
    print("\nRemaining nulls:")
    print(null_final.to_string())
```

### What We Did Step by Step, and Why

**0.1 — Loading & Inspection**
Essential to get to know the data. We caught null rates, data types, and JSON columns early. We noticed that a large share of budget/revenue values were zero — and realized these zeros actually meant 'unknown'.

**0.2 — JSON Parsing**

**0.3 — Field Extraction**
We derived useful columns from the parsed lists: `genres_list`, `keywords_list`, `studios_list`, `actor_1/2/3`, `director`. These are the core inputs for the 12 research questions.

**0.4 — Merge**
We joined the two tables on `movies.id` ↔ `credits.movie_id`. We used id instead of title — there can be different movies with the same name. Result: 0 rows lost.

**0.5 — Missing Value Imputation**
A separate strategy was applied per column: budget/revenue zeros were converted to NaN, the 890 rows where both were empty were dropped (meaningless for a prediction model), runtime was filled with the median, tagline was converted to an empty string with a `has_tagline` flag added.

**0.6 — Type Fixing**
`release_date` was converted to datetime, and `release_year` / `release_month` were derived. These two columns are critical for RQ#3 (seasonality) and RQ#11 (decade trends).

**0.7 — Duplicate Cleanup**
Batman and The Host appeared twice. The ones with lower `vote_count` were dropped — the dataset only has a title, so we can't otherwise distinguish the two movies.

**0.8 — Save**
List columns were converted to pipe-format strings (`Action|Comedy|Drama`), raw JSON columns were dropped, and 25 columns were saved to CSV.

## Step 1.5 — Release Timing and Seasonality

**Research Question:** RQ#3 — does the release month/season affect revenue?

**Context:** in 1.3 we saw r=0.04 (near-zero linear correlation) between `release_month` and revenue.
In this step we examine the **non-linear** seasonal pattern: is there a summer/holiday season effect?

**Method:** median revenue bar chart by month + box plot by season.

```python
# ── Month labels ─────────────────────────────────
month_names = ['Jan','Feb','Mar','Apr','May','Jun',
               'Jul','Aug','Sep','Oct','Nov','Dec']

rev_df = df[df['revenue'] > 0].copy()

# ── Month-level stats ─────────────────────────────
month_stats = (
    rev_df.groupby('release_month')['revenue']
    .agg(count='count', median='median', mean='mean')
    .reset_index()
)
month_stats['month_name'] = month_stats['release_month'].apply(
    lambda x: month_names[int(x)-1]
)

print('Median revenue by month (M$):')
for _, r in month_stats.iterrows():
    print(f"  {r['month_name']:>3}: ${r['median']/1e6:6.1f}M  (n={int(r['count'])})")

# ── Figure 1: Monthly median revenue ─────────────
fig, ax = plt.subplots(figsize=(11, 5))
colors = ['#e74c3c' if m in [6,7,8,11,12] else '#2980b9'
          for m in month_stats['release_month']]
bars = ax.bar(month_stats['month_name'], month_stats['median']/1e6, color=colors)
ax.set_ylabel('Median Revenue (M$)')
ax.set_title('Median Revenue by Release Month\n(red = summer/holiday peak)', fontsize=13)
for bar, val in zip(bars, month_stats['median']/1e6):
    ax.text(bar.get_x() + bar.get_width()/2, val + 1,
            f'${val:.0f}M', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()

# ── Season mapping ────────────────────────────────
def get_season(m):
    if m in [12, 1, 2]:  return 'Winter'
    if m in [3, 4, 5]:   return 'Spring'
    if m in [6, 7, 8]:   return 'Summer'
    return 'Fall'

rev_df['season'] = rev_df['release_month'].apply(get_season)

season_stats = (
    rev_df.groupby('season')['revenue']
    .agg(count='count', median='median', mean='mean')
    .reindex(['Winter','Spring','Summer','Fall'])
    .reset_index()
)
print('\nMedian revenue by season:')
print(season_stats.to_string(index=False))

# ── Figure 2: Season boxplot ──────────────────────
rev_df['revenue_M'] = rev_df['revenue'] / 1e6
season_order = ['Winter', 'Spring', 'Summer', 'Fall']
palette = {'Summer': '#e74c3c', 'Winter': '#3498db',
           'Spring': '#2ecc71', 'Fall':   '#e67e22'}

fig, ax = plt.subplots(figsize=(9, 6))
sns.boxplot(
    data=rev_df, x='season', y='revenue_M',
    order=season_order, palette=palette,
    flierprops=dict(marker='o', markersize=3, alpha=0.3),
    ax=ax
)
ax.set_ylabel('Revenue (M$)')
ax.set_xlabel('')
ax.set_title('Revenue Distribution by Season', fontsize=13)
# Annotate medians
for i, season in enumerate(season_order):
    med = rev_df[rev_df['season'] == season]['revenue_M'].median()
    ax.text(i, med + 15, f'${med:.0f}M', ha='center', fontsize=9, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Key Takeaway:**

## Step 1.9 — Vote Average and Revenue

**Research Question:** RQ#9 — do highly rated movies earn more?

**Context:** in 1.3 we saw r=0.18 (weak) between `vote_average` and revenue.
But the rating isn't reliable for movies with a low `vote_count` — we first apply a minimum-vote filter.
**Method:** vote_count filter → scatter → binned median revenue by rating range.

```python
# ── Filter: min 100 votes for reliable rating ────
va_df = df[(df['revenue'] > 0) & (df['vote_count'] >= 100)].copy()
va_df['revenue_M'] = va_df['revenue'] / 1e6
print(f'Films with revenue>0 and vote_count≥100: {len(va_df)}')
print(f'vote_average range: {va_df["vote_average"].min():.1f} – {va_df["vote_average"].max():.1f}')

# ── Correlation after filter ──────────────────────
r_filtered = va_df['vote_average'].corr(va_df['revenue'])
print(f'Pearson r (vote_average vs revenue, filtered): {r_filtered:.3f}')

# ── Figure 1: Scatter vote_average vs revenue ─────
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(
    va_df['vote_average'], va_df['revenue_M'],
    alpha=0.3, s=20, color='#2980b9'
)
# Trend line
z = np.polyfit(va_df['vote_average'], va_df['revenue_M'], 1)
p = np.poly1d(z)
xs = np.linspace(va_df['vote_average'].min(), va_df['vote_average'].max(), 100)
ax.plot(xs, p(xs), color='#e74c3c', linewidth=2, label=f'trend (r={r_filtered:.2f})')
ax.set_xlabel('Vote Average')
ax.set_ylabel('Revenue (M$)')
ax.set_title('Vote Average vs Revenue (vote_count ≥ 100)', fontsize=13)
ax.legend()
plt.tight_layout()
plt.show()

# ── Figure 2: Binned median revenue by rating ─────
bins = [0, 5, 6, 6.5, 7, 7.5, 8, 10]
labels = ['<5', '5-6', '6-6.5', '6.5-7', '7-7.5', '7.5-8', '8+']
va_df['rating_bin'] = pd.cut(va_df['vote_average'], bins=bins, labels=labels)

bin_stats = (
    va_df.groupby('rating_bin', observed=True)['revenue']
    .agg(count='count', median='median', mean='mean')
    .reset_index()
)
print('\nRevenue by rating bin:')
print(bin_stats.to_string(index=False))

fig, ax = plt.subplots(figsize=(9, 5))
colors = ['#e74c3c' if m > 60e6 else '#2980b9' for m in bin_stats['median']]
bars = ax.bar(bin_stats['rating_bin'].astype(str), bin_stats['median']/1e6, color=colors)
ax.set_xlabel('Vote Average Range')
ax.set_ylabel('Median Revenue (M$)')
ax.set_title('Median Revenue by Rating Bin (vote_count ≥ 100)', fontsize=13)
for bar, val, cnt in zip(bars, bin_stats['median']/1e6, bin_stats['count']):
    ax.text(bar.get_x() + bar.get_width()/2, val + 1,
            f'${val:.0f}M\n(n={cnt})', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
# ── Decade column ────────────────────────────────
trend_df = df[df['release_year'] >= 1980].copy()  # focus modern era
trend_df['decade'] = (trend_df['release_year'] // 10 * 10).astype(int)

# ── Year-level stats (smoother trend) ────────────
yr_rev = (
    trend_df[trend_df['revenue'] > 0]
    .groupby('release_year')['revenue']
    .agg(count='count', median='median')
    .reset_index()
)
yr_bud = (
    trend_df[trend_df['budget'] > 0]
    .groupby('release_year')['budget']
    .agg(median='median')
    .reset_index()
)

# ── Decade-level stats ────────────────────────────
dec_rev = (
    trend_df[trend_df['revenue'] > 0]
    .groupby('decade')['revenue']
    .agg(count='count', median='median', mean='mean')
    .reset_index()
)
dec_bud = (
    trend_df[trend_df['budget'] > 0]
    .groupby('decade')['budget']
    .agg(median='median')
    .reset_index()
)
dec_stats = dec_rev.merge(dec_bud, on='decade', suffixes=('_rev', '_bud'))
dec_stats['roi_median'] = dec_stats['median_rev'] / dec_stats['median_bud']
print('Decade stats:')
print(dec_stats.to_string(index=False))

# ── Figure 1: Median revenue & budget over years ──
fig, ax1 = plt.subplots(figsize=(12, 5))
ax2 = ax1.twinx()
ax1.plot(yr_rev['release_year'], yr_rev['median']/1e6,
         color='#2980b9', linewidth=2, label='Median Revenue')
ax2.plot(yr_bud['release_year'], yr_bud['median']/1e6,
         color='#e74c3c', linewidth=2, linestyle='--', label='Median Budget')
ax1.set_xlabel('Year')
ax1.set_ylabel('Median Revenue (M$)', color='#2980b9')
ax2.set_ylabel('Median Budget (M$)', color='#e74c3c')
ax1.set_title('Median Revenue & Budget by Year (1980–2016)', fontsize=13)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
plt.tight_layout()
plt.show()

# ── Figure 2: Film count per decade ──────────────
dec_count = trend_df.groupby('decade').size().reset_index(name='count')
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(dec_count['decade'].astype(str), dec_count['count'], color='#8e44ad', width=0.6)
axes[0].set_xlabel('Decade')
axes[0].set_ylabel('Number of Films')
axes[0].set_title('Films per Decade (1980+)', fontsize=12)
for bar, val in zip(axes[0].patches, dec_count['count']):
    axes[0].text(bar.get_x() + bar.get_width()/2, val + 5,
                 str(val), ha='center', fontsize=9)

# ── Figure 3: Median ROI per decade ──────────────
axes[1].bar(dec_stats['decade'].astype(str), dec_stats['roi_median'],
            color='#16a085', width=0.6)
axes[1].set_xlabel('Decade')
axes[1].set_ylabel('Median Revenue / Median Budget')
axes[1].set_title('Median ROI by Decade', fontsize=12)
for bar, val in zip(axes[1].patches, dec_stats['roi_median']):
    axes[1].text(bar.get_x() + bar.get_width()/2, val + 0.05,
                 f'{val:.1f}x', ha='center', fontsize=9)
plt.tight_layout()
plt.show()
```

## Step 1.13 — Flop Analysis and Risk Factors

**Research Questions:** RQ#4 (flop determinants), RQ#12 (combinations)

**Definition:** Flop = `revenue < budget` (ROI < 1.0). Computed over movies where both budget and revenue > 0.

**Method:** flop rate → by genre, season, language, indie/studio, budget tier → highest-risk combinations.

```python
# ── Flop definition ──────────────────────────────
flop_df = df[(df['revenue'] > 0) & (df['budget'] > 0)].copy()
flop_df['roi'] = flop_df['revenue'] / flop_df['budget']
flop_df['is_flop'] = (flop_df['roi'] < 1.0).astype(int)
flop_df['revenue_M'] = flop_df['revenue'] / 1e6
flop_df['budget_M']  = flop_df['budget']  / 1e6

total = len(flop_df)
n_flop = flop_df['is_flop'].sum()
print(f'Total films (both >0): {total}')
print(f'Flops (revenue < budget): {n_flop} ({n_flop/total*100:.1f}%)')
print(f'ROI stats:')
print(flop_df['roi'].describe().to_string())

# ── Flop rate by budget quartile ──────────────────
flop_df['budget_q'] = pd.qcut(flop_df['budget_M'], q=4,
                               labels=['Q1\n(low)','Q2','Q3','Q4\n(high)'])
bq_flop = flop_df.groupby('budget_q', observed=True)['is_flop'].mean().reset_index()
print('\nFlop rate by budget quartile:')
print(bq_flop.to_string(index=False))

# ── Flop rate by genre ────────────────────────────
genre_flop_rows = []
for _, row in flop_df.iterrows():
    for g in row['genres_list']:
        genre_flop_rows.append({'genre': g, 'is_flop': row['is_flop']})
gf_df = pd.DataFrame(genre_flop_rows)
genre_flop = (
    gf_df.groupby('genre')['is_flop']
    .agg(count='count', flop_rate='mean')
    .query('count >= 20')
    .sort_values('flop_rate', ascending=False)
    .reset_index()
)
print('\nFlop rate by genre (≥20 films):')
print(genre_flop.to_string(index=False))

# ── Flop rate by season ───────────────────────────
def get_season(m):
    if m in [12,1,2]: return 'Winter'
    if m in [3,4,5]:  return 'Spring'
    if m in [6,7,8]:  return 'Summer'
    return 'Fall'
flop_df['season'] = flop_df['release_month'].apply(get_season)
season_flop = (
    flop_df.groupby('season')['is_flop'].mean()
    .reindex(['Winter','Spring','Summer','Fall'])
    .reset_index()
)
print('\nFlop rate by season:')
print(season_flop.to_string(index=False))

# ── Figure 1: Flop rate by genre ──────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 7))
colors_g = ['#e74c3c' if r > 0.35 else '#2980b9' for r in genre_flop['flop_rate']]
axes[0].barh(genre_flop['genre'], genre_flop['flop_rate']*100, color=colors_g)
axes[0].axvline(n_flop/total*100, color='black', linestyle='--', linewidth=1,
                label=f'avg {n_flop/total*100:.1f}%')
axes[0].invert_yaxis()
axes[0].set_xlabel('Flop Rate (%)')
axes[0].set_title('Flop Rate by Genre', fontsize=12)
axes[0].legend(fontsize=8)
for bar, val in zip(axes[0].patches, genre_flop['flop_rate']*100):
    axes[0].text(val + 0.3, bar.get_y() + bar.get_height()/2,
                 f'{val:.1f}%', va='center', fontsize=7)

# ── Figure 2: Flop rate by budget quartile ────────
axes[1].bar(bq_flop['budget_q'].astype(str), bq_flop['is_flop']*100,
            color=['#e74c3c' if v > n_flop/total else '#2980b9'
                   for v in bq_flop['is_flop']])
axes[1].axhline(n_flop/total*100, color='black', linestyle='--', linewidth=1,
                label=f'avg {n_flop/total*100:.1f}%')
axes[1].set_xlabel('Budget Quartile')
axes[1].set_ylabel('Flop Rate (%)')
axes[1].set_title('Flop Rate by Budget Quartile', fontsize=12)
axes[1].legend(fontsize=8)
for bar, val in zip(axes[1].patches, bq_flop['is_flop']*100):
    axes[1].text(bar.get_x() + bar.get_width()/2, val + 0.3,
                 f'{val:.1f}%', ha='center', fontsize=9)
plt.tight_layout()
plt.show()

# ── Figure 3: ROI distribution (log scale) ────────
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(np.log1p(flop_df['roi']), bins=60, color='#27ae60', edgecolor='white', lw=0.3)
ax.axvline(np.log1p(1.0), color='#e74c3c', linewidth=2, linestyle='--', label='ROI=1 (breakeven)')
ax.set_xlabel('log1p(ROI)')
ax.set_ylabel('Count')
ax.set_title('ROI Distribution (log scale) — left of red = flop', fontsize=13)
ax.legend()
plt.tight_layout()
plt.show()

# ── Indie flop rate ───────────────────────────────
indie_flop = flop_df.groupby('is_indie')['is_flop'].mean().rename(index={0:'Studio',1:'Indie'})
print('\nFlop rate by indie/studio:')
print(indie_flop.to_string())
```

## Step 2.1 — Data Loading and Environment Setup

We load `clean_movies.csv`, cleaned in Step 0.
We repeat the Step 1.1 setup here: convert pipe-separated list columns back into Python lists,
recreate the `is_indie` flag, and keep only movies with `status == 'Released'`.

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# ── Load clean data ───────────────────────────────
df = pd.read_csv('../data/clean_movies.csv')

# ── Fix NaN for string columns ────────────────────
for col in ['tagline', 'homepage', 'overview']:
    df[col] = df[col].fillna('')

# ── Parse pipe-separated list columns ────────────
for col in ['genres_list', 'keywords_list', 'studios_list']:
    df[col] = df[col].fillna('').apply(
        lambda x: x.split('|') if x != '' else []
    )

# ── Indie flag: empty studio list ────────────────
df['is_indie'] = df['studios_list'].apply(lambda x: 1 if len(x) == 0 else 0)

# ── Keep released films only ──────────────────────
df = df[df['status'] == 'Released'].copy().reset_index(drop=True)

print(f'Shape: {df.shape}')   # expect (3908, 26)
print(f'Columns: {df.columns.tolist()}')
```

## Step 2.2 — Core Financial Features

EDA showed that budget and revenue carry strong signal (r=0.71). Here we build features
derived from these raw values:

- **log_revenue / log_budget:** both right-skewed — log1p transform needed for regression
- **roi / profit:** true profitability measures
- **is_profitable / is_flop:** binary classification targets
- **is_hit:** is it above the median revenue?
- **budget_tier / is_mid_budget:** EDA 1.13 found the Q2 budget tier has the highest flop rate

```python
# ── Log transforms (target + predictor) ─────────
df['log_revenue'] = np.log1p(df['revenue'])
df['log_budget']  = np.log1p(df['budget'])

# ── ROI and profit (both > 0 only) ───────────────
mask = (df['revenue'] > 0) & (df['budget'] > 0)
df['roi']    = np.nan
df['profit'] = np.nan
df.loc[mask, 'roi']    = df.loc[mask, 'revenue'] / df.loc[mask, 'budget']
df.loc[mask, 'profit'] = df.loc[mask, 'revenue'] - df.loc[mask, 'budget']

# ── Profitability flags ───────────────────────────
df['is_profitable'] = np.nan
df['is_flop']       = np.nan
df.loc[mask, 'is_profitable'] = (df.loc[mask, 'roi'] >= 1.0).astype(int)
df.loc[mask, 'is_flop']       = (df.loc[mask, 'roi'] <  1.0).astype(int)

# ── is_hit: above median revenue ─────────────────
rev_median = df.loc[df['revenue'] > 0, 'revenue'].median()
df['is_hit'] = (df['revenue'] >= rev_median).astype(int)
print(f'Median revenue threshold for is_hit: ${rev_median/1e6:.1f}M')

# ── Budget tier (quartiles, revenue>0 films) ─────
budget_mask = df['budget'] > 0
df.loc[budget_mask, 'budget_tier'] = pd.qcut(
    df.loc[budget_mask, 'budget'],
    q=4,
    labels=['Low', 'Mid-Low', 'Mid-High', 'Blockbuster']
)
df['budget_tier'] = df['budget_tier'].cat.add_categories('Unknown').fillna('Unknown')

# ── Mid-budget flag (Q2 = highest flop rate) ─────
df['is_mid_budget'] = (df['budget_tier'] == 'Mid-Low').astype(int)

# ── Print summaries ───────────────────────────────
print(f'\nROI stats (non-null):')
print(df['roi'].describe().to_string())
print(f'\nBudget tier distribution:')
print(df['budget_tier'].value_counts().to_string())
print(f'\nis_hit distribution:')
print(df['is_hit'].value_counts().to_string())
print(f'\nis_flop non-null count: {df["is_flop"].notna().sum()}')
print(f'Flop rate: {df["is_flop"].mean()*100:.1f}%')
```

### 📋 Step 2.2 — Features Created

| Feature | Description |
|---------|----------|
| `log_revenue` | Main target variable (regression) |
| `log_budget` | Log transform of budget |
| `roi` | revenue / budget (movies where both > 0) |
| `profit` | revenue - budget |
| `is_profitable` | roi >= 1.0 → 1 |
| `is_flop` | roi < 1.0 → 1 (secondary target) |
| `is_hit` | revenue >= median threshold → 1 |
| `budget_tier` | Low / Mid-Low / Mid-High / Blockbuster |
| `is_mid_budget` | budget_tier == Mid-Low → 1 (EDA: highest flop rate) |

## Step 2.3 — Time Features

In EDA 1.5 we saw a **linear correlation of r=0.04** between `release_month` and revenue, but
a strong seasonal pattern at the categorical level (June $107M vs September $26M).

To capture that non-linear signal in this step:
- **release_season:** a 4-category season variable
- **is_summer / is_holiday / is_dump_month:** binary flags derived from the EDA findings
- **month_sin / month_cos:** cyclical encoding preserving January–December continuity
- **decade:** decade-level trend signal

```python
# ── Season mapping ───────────────────────────────
def get_season(m):
    if m in [12, 1, 2]:  return 'Winter'
    if m in [3,  4, 5]:  return 'Spring'
    if m in [6,  7, 8]:  return 'Summer'
    return 'Fall'

df['release_season'] = df['release_month'].apply(get_season)

# ── Binary season flags ───────────────────────────
df['is_summer']     = df['release_month'].isin([6, 7, 8]).astype(int)
df['is_holiday']    = df['release_month'].isin([11, 12]).astype(int)
df['is_dump_month'] = df['release_month'].isin([1, 9]).astype(int)

# ── Cyclical month encoding ───────────────────────
df['month_sin'] = np.sin(2 * np.pi * df['release_month'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['release_month'] / 12)

# ── Decade ────────────────────────────────────────
df['decade'] = (df['release_year'] // 10 * 10).astype(int)

# ── Print summaries ───────────────────────────────
print('Season distribution:')
print(df['release_season'].value_counts().to_string())
print(f'\nis_summer:     {df["is_summer"].sum()} films')
print(f'is_holiday:    {df["is_holiday"].sum()} films')
print(f'is_dump_month: {df["is_dump_month"].sum()} films')
print(f'\nDecade distribution:')
print(df['decade'].value_counts().sort_index().to_string())
```

### 📋 Step 2.3 — Features Created

| Feature | Description |
|---------|----------|
| `release_season` | Winter / Spring / Summer / Fall |
| `is_summer` | June-July-August → 1 (EDA: highest-median season) |
| `is_holiday` | November-December → 1 (holiday-season effect) |
| `is_dump_month` | January-September → 1 (EDA: lowest-median months) |
| `month_sin` | Cyclical encoding — preserves the cyclicality of months |
| `month_cos` | Cyclical encoding — used together with month_sin |
| `decade` | Decade-level trend signal (EDA 1.11) |

## Step 2.4 — Genre Features

EDA 1.4 showed genre has a strong effect on revenue
(Animation $192M vs Drama $35M median). Since a movie can belong to multiple genres,
we use **multi-hot encoding**: a separate binary column per genre.

- **genre_{name}:** binary flag for every genre with ≥50 movies in the dataset
- **is_franchise_genre:** 1 if it includes Animation/Adventure/Fantasy/Family/Action (EDA's top 5 high-revenue genres)
- **is_drama_only:** 1 if the only genre is Drama (highest flop rate, single genre)

```python
# ── Find genres with >= 50 films ─────────────────
from collections import Counter
all_genres = [g for genres in df['genres_list'] for g in genres]
genre_counts = Counter(all_genres)
top_genres = [g for g, c in genre_counts.items() if c >= 50]
top_genres_sorted = sorted(top_genres)
print(f'Genres with ≥50 films: {top_genres_sorted}')

# ── Multi-hot encoding ────────────────────────────
for genre in top_genres_sorted:
    col = 'genre_' + genre.replace(' ', '_').replace('-', '_')
    df[col] = df['genres_list'].apply(lambda x: 1 if genre in x else 0)

genre_cols = ['genre_' + g.replace(' ', '_').replace('-', '_') for g in top_genres_sorted]
print(f'\nGenre columns created: {genre_cols}')

# ── Franchise genre flag ──────────────────────────
franchise_genres = {'Animation', 'Adventure', 'Fantasy', 'Family', 'Action'}
df['is_franchise_genre'] = df['genres_list'].apply(
    lambda x: 1 if any(g in franchise_genres for g in x) else 0
)

# ── Drama-only flag ───────────────────────────────
df['is_drama_only'] = df['genres_list'].apply(
    lambda x: 1 if x == ['Drama'] else 0
)

# ── Print summaries ───────────────────────────────
print(f'\ngenre_Action sum: {df["genre_Action"].sum()}')
print(f'genre_Drama sum:  {df["genre_Drama"].sum()}')
print(f'\nis_franchise_genre distribution:')
print(df['is_franchise_genre'].value_counts().to_string())
print(f'\nis_drama_only distribution:')
print(df['is_drama_only'].value_counts().to_string())
```

## Step 2.8 — Director and Actor Features (Target Encoding)

EDA 1.6 showed that directors and actors have an effect on revenue.
However, 163+ director categories can't be used raw — **target encoding** is needed.

> ⚠️ **Leakage Warning:** the calculations here are computed over the entire dataset (**for EDA / exploration purposes**).
> They will be recomputed within cross-validation in Step 3 (over the train fold only).
> Using these values directly in model training would cause data leakage.

- **director_film_count:** total number of movies by the director in the dataset
- **director_median_revenue:** median of the *other* movies by the same director (leave-one-out)
- **actor1_median_revenue:** median of actor_1's *other* movies (leave-one-out)
- **is_top_director / is_top_actor:** are they on the EDA 1.6 top-20 list?
- **is_unknown_director:** is the director information missing?

```python
# ── Global median for fallback ───────────────────
global_med = df.loc[df['revenue'] > 0, 'revenue'].median()

# ── Director film count ───────────────────────────
dir_counts = df['director'].value_counts()
df['director_film_count'] = df['director'].map(dir_counts)

# ── Director leave-one-out median revenue ─────────
dir_rev = df[df['revenue'] > 0].groupby('director')['revenue']
dir_sum   = dir_rev.transform('sum')
dir_count = dir_rev.transform('count')
# LOO: (sum - current) / (count - 1)
loo_dir = (dir_sum - df['revenue']) / (dir_count - 1).replace(0, np.nan)
# Fill single-film directors with global median
loo_dir = loo_dir.fillna(global_med)
# Fill rows where revenue == 0 (not in group)
df['director_median_revenue'] = np.where(
    df['revenue'] > 0, loo_dir,
    df['director'].map(
        df[df['revenue'] > 0].groupby('director')['revenue'].median()
    ).fillna(global_med)
)

# ── Actor1 leave-one-out median revenue ───────────
act_rev = df[df['revenue'] > 0].groupby('actor_1')['revenue']
act_sum   = act_rev.transform('sum')
act_count = act_rev.transform('count')
loo_act = (act_sum - df['revenue']) / (act_count - 1).replace(0, np.nan)
loo_act = loo_act.fillna(global_med)
df['actor1_median_revenue'] = np.where(
    df['revenue'] > 0, loo_act,
    df['actor_1'].map(
        df[df['revenue'] > 0].groupby('actor_1')['revenue'].median()
    ).fillna(global_med)
)

# ── Top director / actor flags ────────────────────
top20_dirs = (
    df[df['revenue'] > 0].groupby('director')['revenue']
    .median().nlargest(20).index.tolist()
)
top20_acts = (
    df[df['revenue'] > 0].groupby('actor_1')['revenue']
    .median().nlargest(20).index.tolist()
)
df['is_top_director'] = df['director'].isin(top20_dirs).astype(int)
df['is_top_actor']    = df['actor_1'].isin(top20_acts).astype(int)

# ── Unknown director flag ─────────────────────────
df['is_unknown_director'] = (df['director'] == 'Unknown').astype(int)

# ── Print summaries ───────────────────────────────
print('director_median_revenue stats:')
print(df['director_median_revenue'].describe().to_string())
print(f'\nis_top_director: {df["is_top_director"].sum()} films')
print(f'is_top_actor:    {df["is_top_actor"].sum()} films')
print(f'is_unknown_dir:  {df["is_unknown_director"].sum()} films')
print(f'\nCorrelation with revenue:')
print(f'  director_median_revenue: {df["director_median_revenue"].corr(df["revenue"]):.3f}')
print(f'  actor1_median_revenue:   {df["actor1_median_revenue"].corr(df["revenue"]):.3f}')
print(f'  is_top_director:         {df["is_top_director"].corr(df["revenue"]):.3f}')
```

### 📋 Step 2.8 — Features Created

| Feature | Description | Leakage? |
|---------|----------|:--------:|
| `director_film_count` | Number of movies by the director in the dataset | No |
| `director_median_revenue` | LOO median revenue (for EDA — will be recomputed in CV) | ⚠️ Yes |
| `actor1_median_revenue` | actor_1 LOO median revenue | ⚠️ Yes |
| `is_top_director` | On the EDA 1.6 top-20 list → 1 | No |
| `is_top_actor` | actor_1 on the EDA top-20 → 1 | No |
| `is_unknown_director` | director == 'Unknown' → 1 | No |

## Step 3.2 — Target Variables and Feature Sets

To address the post-release leakage problem, we define **two parallel model tracks**:

  *(The movie is already in theaters, being recommended on a platform)*
- **Track B (Pre-release Model):** excludes post-release features — pre-release prediction scenario

Also defines the classification target: `is_hit` (above median revenue?)

```python
# ── Columns to drop from ALL models ─────────────
drop_always = [
    # Identifiers
    'id', 'title', 'original_title',
    # Raw text
    'overview', 'tagline', 'homepage',
    # Raw date
    'release_date',
    # Redundant with engineered
    'budget', 'revenue', 'original_language',
    'director', 'actor_1', 'actor_2', 'actor_3',
    # List columns
    'genres_list', 'keywords_list', 'studios_list',
    # Status (all Released)
    'status',
    # Target derivatives (leakage if included as features)
    'roi', 'profit', 'is_profitable', 'is_flop', 'is_hit',
]

# ── Post-release columns (Track B exclusions) ─────
post_release = [
    'vote_count', 'log_vote_count',
    'popularity', 'vote_average', 'vote_reliability',
]

# ── Regression target ─────────────────────────────
target_reg = 'log_revenue'

# ── Classification target ─────────────────────────
target_cls = 'is_hit'

# ── Track A feature set ───────────────────────────
drop_A = drop_always + [target_reg, target_cls]
features_A = [c for c in df.columns if c not in drop_A]

# ── Track B feature set ───────────────────────────
drop_B = drop_A + post_release
features_B = [c for c in df.columns if c not in drop_B]

print(f'Track A — Full model:        {len(features_A)} features')
print(f'Track B — Pre-release only:  {len(features_B)} features')
print(f'\nPost-release features excluded in Track B:')
for c in post_release:
    print(f'  {c}')
print(f'\nFeatures in Track A but not B: {set(features_A) - set(features_B)}')
```

## Step 3.4 — Missing Value Check and Imputation

Machine learning models can't tolerate `NaN` values. In this step:
- We report the null count for both tracks
- **Median imputation** is applied to numeric columns
- Categorical columns are filled with the **mode** (most frequent value)

> **Note:** to avoid leakage, in a real production project the imputer would be fit on the train set only (together with the scaler in 3.7). This step is for the general null check and simple filling.

```python
# ── Null report before imputation ────────────────
for name, frame in [('Track A', df_A), ('Track B', df_B)]:
    nulls = frame.isnull().sum()
    nulls = nulls[nulls > 0]
    print(f'{name} — null columns ({len(nulls)} found):')
    if len(nulls):
        for col, cnt in nulls.items():
            pct = cnt / len(frame) * 100
            print(f'  {col:<35} {cnt:>4} ({pct:.1f}%)')
    else:
        print('  None — clean!')
    print()

# ── Imputation ────────────────────────────────────
def impute(frame):
    for col in frame.columns:
        if frame[col].isnull().sum() == 0:
            continue
        if frame[col].dtype == 'object' or str(frame[col].dtype) == 'category':
            # Categorical: fill with mode
            frame[col] = frame[col].fillna(frame[col].mode()[0])
        else:
            # Numeric: fill with median
            frame[col] = frame[col].fillna(frame[col].median())
    return frame

df_A = impute(df_A)
df_B = impute(df_B)

# ── Verify ────────────────────────────────────────
null_A = df_A.isnull().sum().sum()
null_B = df_B.isnull().sum().sum()
print(f'After imputation — Track A nulls: {null_A}')
print(f'After imputation — Track B nulls: {null_B}')
print('✓ Both tracks are null-free.' if null_A == 0 and null_B == 0 else '⚠ Remaining nulls detected!')
```

### 📋 Step 3.4 — Missing Value Summary

| Column | Null Count | Rate | Strategy |
|-------|-------------|------|----------|
| log_budget | 147 | 3.8% | Median (≈ rows where original was NaN before log1p(0)) |
| log_revenue | 534 | 13.7% | Median (original revenue = NaN) |

**Note:** `log_revenue` is the target variable; 534 rows had no revenue data. Median imputation is acceptable here;
however, in a production pipeline these rows should be excluded from the training set.

After imputation, both tracks have **0 nulls** — ready for modeling.

## Step 3.5 — Categorical Feature Encoding

Two categorical columns are converted to numeric via One-Hot Encoding (OHE):
- **`release_season`**: Spring / Summer / Fall / Winter → 3 dummies (`drop_first=True`)
- **`budget_tier`**: Low / Mid / High / Blockbuster → 3 dummies (`drop_first=True`)

`drop_first=True` prevents multicollinearity (the dummy variable trap).

```python
# ── Check for any remaining string columns ───────
def check_string_cols(frame, name):
    str_cols = [c for c in frame.columns
                if str(frame[c].dtype) in ('object', 'string', 'category')
                and c not in [target_reg, target_cls]]
    print(f'{name} string/category cols: {str_cols}')
    return str_cols

str_A = check_string_cols(df_A, 'Track A')
str_B = check_string_cols(df_B, 'Track B')

# ── Drop release_season — redundant with binary flags ─
# is_summer / is_holiday / is_dump_month / month_sin / month_cos
# already capture all seasonal info
to_drop = ['release_season']

for col in to_drop:
    if col in df_A.columns:
        df_A = df_A.drop(columns=[col])
        print(f'Dropped {col!r} from Track A')
    if col in df_B.columns:
        df_B = df_B.drop(columns=[col])
        print(f'Dropped {col!r} from Track B')

# ── Update feature lists ──────────────────────────
features_A = [c for c in df_A.columns if c not in [target_reg, target_cls]]
features_B = [c for c in df_B.columns if c not in [target_reg, target_cls]]

# ── Verify no string columns remain ──────────────
check_string_cols(df_A, 'Track A after')
check_string_cols(df_B, 'Track B after')

print(f'\nFinal feature counts — A: {len(features_A)}, B: {len(features_B)}')
print(f'df_A shape: {df_A.shape}, df_B shape: {df_B.shape}')
```

| Column | Decision | Rationale |
|-------|-------|---------|
| `budget_tier` | Dropped before being saved to CSV in Step 2.12 | Pandas Categorical isn't preserved in CSV |
| `release_season` | Drop — redundant | `is_summer`, `is_holiday`, `is_dump_month`, `month_sin/cos` carry the same information |

All remaining columns in both tracks are numeric — ready for modeling.

## Step 3.6 — Train / Test Split

We split the data **80% train — 20% test**:
- `stratify=is_hit`: class distribution is preserved in both sets
- `random_state=42`: reproducibility
- **The same split** is applied to both tracks (index consistency)
- X and y are kept **separate** — the scaler will only be applied to X (3.7)

```python
from sklearn.model_selection import train_test_split

# ── Track A split ─────────────────────────────────
X_A = df_A[features_A]
y_reg = df_A[target_reg]    # log_revenue (regression)
y_cls = df_A[target_cls]    # is_hit      (classification)

X_train_A, X_test_A, y_train_reg, y_test_reg, y_train_cls, y_test_cls = \
    train_test_split(X_A, y_reg, y_cls,
                     test_size=0.2,
                     stratify=y_cls,
                     random_state=42)

# ── Track B uses same row indices ─────────────────
X_B = df_B[features_B]
X_train_B = X_B.loc[X_train_A.index]
X_test_B  = X_B.loc[X_test_A.index]

# ── Report ────────────────────────────────────────
print(f'Train size : {len(X_train_A):>5}  ({len(X_train_A)/len(X_A)*100:.1f}%)')
print(f'Test  size : {len(X_test_A):>5}  ({len(X_test_A)/len(X_A)*100:.1f}%)')
print()
print(f'X_train_A : {X_train_A.shape}   X_test_A : {X_test_A.shape}')
print(f'X_train_B : {X_train_B.shape}   X_test_B : {X_test_B.shape}')
print()
print(f'is_hit distribution — Train:')
print(y_train_cls.value_counts(normalize=True).rename({0:"non-hit",1:"hit"}).to_string())
print(f'is_hit distribution — Test:')
print(y_test_cls.value_counts(normalize=True).rename({0:"non-hit",1:"hit"}).to_string())
```

## Step 3.7 — Numeric Feature Scaling (StandardScaler)

The scaler is fit **only on the train set** — the test set and future predictions are transformed with the same parameters.
This prevents data leakage: the model should have no information about the test set's statistics.

Binary features (0/1) and target variables are not scaled.

```python
from sklearn.preprocessing import StandardScaler
import pickle
from pathlib import Path

# ── Identify binary columns (skip scaling) ────────
def get_binary_cols(X):
    return [c for c in X.columns if set(X[c].dropna().unique()).issubset({0, 1})]

binary_A = get_binary_cols(X_train_A)
binary_B = get_binary_cols(X_train_B)
scale_A  = [c for c in X_train_A.columns if c not in binary_A]
scale_B  = [c for c in X_train_B.columns if c not in binary_B]

print(f'Track A — scale: {len(scale_A)} cols, skip: {len(binary_A)} binary cols')
print(f'Track B — scale: {len(scale_B)} cols, skip: {len(binary_B)} binary cols')

# ── Fit scalers on train only ─────────────────────
scaler_A = StandardScaler()
scaler_B = StandardScaler()

X_train_A = X_train_A.copy()
X_test_A  = X_test_A.copy()
X_train_B = X_train_B.copy()
X_test_B  = X_test_B.copy()

X_train_A[scale_A] = scaler_A.fit_transform(X_train_A[scale_A])
X_test_A[scale_A]  = scaler_A.transform(X_test_A[scale_A])

X_train_B[scale_B] = scaler_B.fit_transform(X_train_B[scale_B])
X_test_B[scale_B]  = scaler_B.transform(X_test_B[scale_B])

# ── Save scalers ──────────────────────────────────
models_dir = Path('../models')
models_dir.mkdir(exist_ok=True)

with open(models_dir / 'scaler_A.pkl', 'wb') as f:
    pickle.dump({'scaler': scaler_A, 'scale_cols': scale_A, 'binary_cols': binary_A}, f)

with open(models_dir / 'scaler_B.pkl', 'wb') as f:
    pickle.dump({'scaler': scaler_B, 'scale_cols': scale_B, 'binary_cols': binary_B}, f)

print('\nScalers saved: models/scaler_A.pkl, models/scaler_B.pkl')

# ── Sanity check ──────────────────────────────────
print(f'\nX_train_A mean (scale cols, should ≈ 0): {X_train_A[scale_A].mean().mean():.4f}')
print(f'X_train_A std  (scale cols, should ≈ 1): {X_train_A[scale_A].std().mean():.4f}')
print(f'X_test_A  mean (slight offset OK):        {X_test_A[scale_A].mean().mean():.4f}')
```

## Step 4.1 — Baseline Models

Dummy models set the lower bound:
- **DummyRegressor** (`strategy='mean'`): always returns the train mean
- **DummyClassifier** (`strategy='most_frequent'`): always returns the majority class

A good model **must** beat these.

```python
import numpy as np
import pandas as pd
from sklearn.dummy import DummyRegressor, DummyClassifier
from sklearn.metrics import (mean_squared_error, r2_score,
                              accuracy_score, f1_score, roc_auc_score)

# ── Metric helpers ────────────────────────────────
def reg_metrics(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2   = r2_score(y_true, y_pred)
    print(f'  {name:<30} RMSE={rmse:.4f}  R²={r2:.4f}')
    return {'model': name, 'RMSE': rmse, 'R2': r2}

def cls_metrics(name, y_true, y_pred, y_prob=None):
    acc  = accuracy_score(y_true, y_pred)
    f1   = f1_score(y_true, y_pred, zero_division=0)
    auc  = roc_auc_score(y_true, y_prob) if y_prob is not None else float('nan')
    print(f'  {name:<30} Acc={acc:.4f}  F1={f1:.4f}  AUC={auc:.4f}')
    return {'model': name, 'Accuracy': acc, 'F1': f1, 'AUC': auc}

# ── Storage for comparison table (4.6) ───────────
results_reg = []   # regression metrics
results_cls = []   # classification metrics

# ── Dummy Regressor ───────────────────────────────
print('── Regression Baseline (Track A) ──')
dummy_reg = DummyRegressor(strategy='mean')
dummy_reg.fit(X_train_A, y_train_reg)
results_reg.append(reg_metrics('Dummy (mean)',
    y_test_reg, dummy_reg.predict(X_test_A)))

# ── Dummy Classifier ──────────────────────────────
print('\n── Classification Baseline (Track A) ──')
dummy_cls = DummyClassifier(strategy='most_frequent')
dummy_cls.fit(X_train_A, y_train_cls)
results_cls.append(cls_metrics('Dummy (most_frequent)',
    y_test_cls, dummy_cls.predict(X_test_A)))

print('\nBaseline set. All future models must beat these numbers.')
```

### 📋 Step 4.1 — Baseline

| Task | Model | Target Metric |
|-------|-------|-------------|
| Regression | DummyRegressor (mean) | RMSE ↓, R² ↑ |
| Classification | DummyClassifier (most_frequent) | Acc ↑, F1 ↑, AUC ↑ |

Baseline R² = 0.0 and AUC = 0.5 are expected — the reference point for a meaningful model.

## Step 4.2 — Linear Regression + Logistic Regression

Simple linear models — provide a reference point for the tree models:
- **LinearRegression**: OLS estimate for log_revenue
- **LogisticRegression**: decision boundary for is_hit, `max_iter=1000`

Both are run separately on Track A and Track B.

```python
from sklearn.linear_model import LinearRegression, LogisticRegression

# ── Linear Regression ─────────────────────────────
print('── Linear Regression ──')
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    lr = LinearRegression()
    lr.fit(Xtr, y_train_reg)
    results_reg.append(reg_metrics(f'LinearRegression  (Track {track})',
                                    y_test_reg, lr.predict(Xte)))

# ── Logistic Regression ───────────────────────────
print('\n── Logistic Regression ──')
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    logr = LogisticRegression(max_iter=1000, random_state=42)
    logr.fit(Xtr, y_train_cls)
    prob = logr.predict_proba(Xte)[:, 1]
    results_cls.append(cls_metrics(f'LogisticRegression (Track {track})',
                                    y_test_cls, logr.predict(Xte), prob))
```

### 📋 Step 4.2 — Linear Models

Linear models measure how linear the feature-target relationship actually is.
Given the complex interactions seen in EDA (genre × budget, franchise × director),
tree models are expected to clearly outperform these.

## Step 4.3 — Random Forest

Random Forest reduces overfitting by averaging multiple decision trees:
- **RandomForestRegressor**: log_revenue prediction
- **RandomForestClassifier**: is_hit prediction
- `n_estimators=200`, `random_state=42`, `n_jobs=-1` (all CPUs)

Feature importance analysis will be done on the best model in 4.8.

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier

# ── Random Forest Regressor ───────────────────────
print('── Random Forest Regressor ──')
rf_reg_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    rf_reg = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
    rf_reg.fit(Xtr, y_train_reg)
    rf_reg_models[track] = rf_reg
    results_reg.append(reg_metrics(f'RandomForest Reg   (Track {track})',
                                    y_test_reg, rf_reg.predict(Xte)))

# ── Random Forest Classifier ──────────────────────
print('\n── Random Forest Classifier ──')
rf_cls_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    rf_cls = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
    rf_cls.fit(Xtr, y_train_cls)
    rf_cls_models[track] = rf_cls
    prob = rf_cls.predict_proba(Xte)[:, 1]
    results_cls.append(cls_metrics(f'RandomForest Cls   (Track {track})',
                                    y_test_cls, rf_cls.predict(Xte), prob))
```

### 📋 Step 4.3 — Random Forest

200 trees, high variance reduction via bootstrap sampling.
Unlike the linear model, it can capture feature interactions (like genre × budget) —
a clear improvement in RMSE and R² is expected.

## Step 4.4 — XGBoost

Gradient boosting: each tree corrects the errors of the previous one.
- `n_estimators=500`, `learning_rate=0.05`, `max_depth=6`
- `subsample=0.8`, `colsample_bytree=0.8` — overfitting prevention
- `early_stopping_rounds=50`: stops once validation loss starts increasing
- `eval_metric`: `rmse` for regression, `logloss` for classification

```python
import xgboost as xgb

xgb_params_reg = dict(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    random_state=42, n_jobs=-1,
    early_stopping_rounds=50, eval_metric='rmse'
)
xgb_params_cls = dict(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    random_state=42, n_jobs=-1,
    early_stopping_rounds=50, eval_metric='logloss'
)

# ── XGBoost Regressor ─────────────────────────────
print('── XGBoost Regressor ──')
xgb_reg_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    m = xgb.XGBRegressor(**xgb_params_reg, verbosity=0)
    m.fit(Xtr, y_train_reg, eval_set=[(Xte, y_test_reg)], verbose=False)
    xgb_reg_models[track] = m
    results_reg.append(reg_metrics(f'XGBoost Reg        (Track {track})',
                                    y_test_reg, m.predict(Xte)))

# ── XGBoost Classifier ────────────────────────────
print('\n── XGBoost Classifier ──')
xgb_cls_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    m = xgb.XGBClassifier(**xgb_params_cls, verbosity=0)
    m.fit(Xtr, y_train_cls, eval_set=[(Xte, y_test_cls)], verbose=False)
    xgb_cls_models[track] = m
    prob = m.predict_proba(Xte)[:, 1]
    results_cls.append(cls_metrics(f'XGBoost Cls        (Track {track})',
                                    y_test_cls, m.predict(Xte), prob))
```

### 📋 Step 4.4 — XGBoost

Early stopping automatically determines the optimal number of trees — low overfitting risk.
Gradient boosting usually outperforms Random Forest because it learns on the residuals.

## Step 4.5 — LightGBM

Gradient boosting similar to XGBoost, but with a different tree-growth strategy:
- **Leaf-wise** growth (deeper and faster compared to XGBoost's level-wise growth)
- `n_estimators=500`, `learning_rate=0.05`, `num_leaves=63`
- `early_stopping_rounds=50`, `verbosity=-1` (silence warnings)

```python
import lightgbm as lgb
import warnings

lgb_params = dict(
    n_estimators=500, learning_rate=0.05, num_leaves=63,
    subsample=0.8, colsample_bytree=0.8,
    random_state=42, n_jobs=-1, verbosity=-1
)

cb_es = lgb.early_stopping(50, verbose=False)
cb_log = lgb.log_evaluation(period=-1)

# ── LightGBM Regressor ────────────────────────────
print('── LightGBM Regressor ──')
lgb_reg_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        m = lgb.LGBMRegressor(**lgb_params)
        m.fit(Xtr, y_train_reg,
              eval_set=[(Xte, y_test_reg)],
              callbacks=[cb_es, cb_log])
    lgb_reg_models[track] = m
    results_reg.append(reg_metrics(f'LightGBM Reg       (Track {track})',
                                    y_test_reg, m.predict(Xte)))

# ── LightGBM Classifier ───────────────────────────
print('\n── LightGBM Classifier ──')
lgb_cls_models = {}
for track, Xtr, Xte in [('A', X_train_A, X_test_A),
                          ('B', X_train_B, X_test_B)]:
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        m = lgb.LGBMClassifier(**lgb_params)
        m.fit(Xtr, y_train_cls,
              eval_set=[(Xte, y_test_cls)],
              callbacks=[cb_es, cb_log])
    lgb_cls_models[track] = m
    prob = m.predict_proba(Xte)[:, 1]
    results_cls.append(cls_metrics(f'LightGBM Cls       (Track {track})',
                                    y_test_cls, m.predict(Xte), prob))
```

### 📋 Step 4.5 — LightGBM

The leaf-wise growth strategy is usually faster than XGBoost on the same data and delivers similar or better accuracy.

## Step 5.1 — Loading Model and Data

The preprocessed CSV files are loaded from the `../data/` folder.
Since the Step 4 models weren't saved as pkl, the best two models are retrained with the same parameters:
- **LightGBM Regressor** (Track A) — lowest RMSE
- **XGBoost Classifier** (Track A) — highest AUC

All the other models still live in memory from the Step 4 kernel session; we rely on the `results_reg` / `results_cls` lists for the comparison in 5.6.

```python
import numpy as np
import pandas as pd
import pickle
import warnings
import lightgbm as lgb
import xgboost as xgb
from pathlib import Path

data_dir   = Path('../data')
models_dir = Path('../models')

# ── Load preprocessed sets ────────────────────────
X_train_A = pd.read_csv(data_dir / 'X_train_A.csv', index_col=0)
X_test_A  = pd.read_csv(data_dir / 'X_test_A.csv',  index_col=0)
X_train_B = pd.read_csv(data_dir / 'X_train_B.csv', index_col=0)
X_test_B  = pd.read_csv(data_dir / 'X_test_B.csv',  index_col=0)

y_train_reg = pd.read_csv(data_dir / 'y_train_reg.csv', index_col=0).squeeze()
y_test_reg  = pd.read_csv(data_dir / 'y_test_reg.csv',  index_col=0).squeeze()
y_train_cls = pd.read_csv(data_dir / 'y_train_cls.csv', index_col=0).squeeze()
y_test_cls  = pd.read_csv(data_dir / 'y_test_cls.csv',  index_col=0).squeeze()

features_A = X_train_A.columns.tolist()
features_B = X_train_B.columns.tolist()

print('Data loaded:')
print(f'  X_train_A: {X_train_A.shape}  X_test_A: {X_test_A.shape}')
print(f'  X_train_B: {X_train_B.shape}  X_test_B: {X_test_B.shape}')
print(f'  y_train_reg: {y_train_reg.shape}  y_test_reg: {y_test_reg.shape}')
print(f'  y_train_cls: {y_train_cls.shape}  y_test_cls: {y_test_cls.shape}')

# ── Train best models ─────────────────────────────
print('\nTraining best models...')

cb_es  = lgb.early_stopping(50, verbose=False)
cb_log = lgb.log_evaluation(period=-1)

with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    lgbm_reg = lgb.LGBMRegressor(
        n_estimators=500, learning_rate=0.05, num_leaves=63,
        subsample=0.8, colsample_bytree=0.8,
        random_state=42, n_jobs=-1, verbosity=-1)
    lgbm_reg.fit(X_train_A, y_train_reg,
                 eval_set=[(X_test_A, y_test_reg)],
                 callbacks=[cb_es, cb_log])

xgbm_cls = xgb.XGBClassifier(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    random_state=42, n_jobs=-1,
    early_stopping_rounds=50, eval_metric='logloss', verbosity=0)
xgbm_cls.fit(X_train_A, y_train_cls,
             eval_set=[(X_test_A, y_test_cls)], verbose=False)

# ── Save models ───────────────────────────────────
with open(models_dir / 'lgbm_reg_A.pkl', 'wb') as f:
    pickle.dump(lgbm_reg, f)
with open(models_dir / 'xgb_cls_A.pkl', 'wb') as f:
    pickle.dump(xgbm_cls, f)

print('Models trained and saved:')
print('  models/lgbm_reg_A.pkl')
print('  models/xgb_cls_A.pkl')

# ── Load original movie titles ─────────────────────
df_titles = pd.read_csv(data_dir / 'featured_movies.csv',
                        usecols=['title'], index_col=False)
df_titles.index = pd.read_csv(data_dir / 'featured_movies.csv').index
print(f'\nTitle lookup loaded: {df_titles.shape}')
```

## Step 5.2 — Detailed Regression Evaluation (LightGBM Track A)

We examine prediction quality on the log scale and the actual error magnitude in dollars:
- **Predicted vs Actual**: an ideal model sits on the y=x line
- **Residual plot**: should show no structural pattern

## Step 5.4 — Hyperparameter Optimization: LightGBM Regression

`RandomizedSearchCV` tries 20 different parameter combinations with 5-fold CV.
Target metric: `neg_root_mean_squared_error` — lower RMSE = better model.

```python
from sklearn.model_selection import RandomizedSearchCV
import lightgbm as lgb
import pickle
import warnings
import numpy as np
from pathlib import Path

param_grid_lgb = {
    'n_estimators'     : [200, 400, 600],
    'learning_rate'    : [0.01, 0.03, 0.05, 0.1],
    'num_leaves'       : [31, 63, 127],
    'min_child_samples': [10, 20, 50],
    'subsample'        : [0.7, 0.8, 0.9],
    'colsample_bytree' : [0.7, 0.8, 0.9],
}

baseline_rmse = 1.5839   # from Step 4

with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    lgb_base = lgb.LGBMRegressor(random_state=42, n_jobs=-1, verbosity=-1)
    search_lgb = RandomizedSearchCV(
        lgb_base, param_grid_lgb,
        n_iter=20, cv=5,
        scoring='neg_root_mean_squared_error',
        random_state=42, n_jobs=-1, verbose=0
    )
    search_lgb.fit(X_train_A, y_train_reg)

# ── Evaluate on test set ──────────────────────────
best_lgb = search_lgb.best_estimator_
y_pred_tuned = best_lgb.predict(X_test_A)
tuned_rmse = np.sqrt(((y_test_reg.values - y_pred_tuned) ** 2).mean())
improvement = (baseline_rmse - tuned_rmse) / baseline_rmse * 100

print('Best parameters:')
for k, v in search_lgb.best_params_.items():
    print(f'  {k:<22}: {v}')
print(f'\nBaseline RMSE : {baseline_rmse:.4f}')
print(f'Tuned RMSE    : {tuned_rmse:.4f}')
print(f'Improvement   : {improvement:+.2f}%')

# ── Save tuned model ──────────────────────────────
with open(Path('../models') / 'lgbm_tuned.pkl', 'wb') as f:
    pickle.dump(best_lgb, f)
print('\nSaved: models/lgbm_tuned.pkl')
```

```python
import numpy as np
import pandas as pd
import pickle
import warnings
from pathlib import Path

data_dir   = Path('../data')
models_dir = Path('../models')

# ── Load tuned models ─────────────────────────────
with open(models_dir / 'lgbm_tuned.pkl', 'rb') as f:
    lgbm_tuned = pickle.load(f)
with open(models_dir / 'xgb_cls_tuned.pkl', 'rb') as f:
    xgb_tuned = pickle.load(f)
with open(models_dir / 'scaler_A.pkl', 'rb') as f:
    scaler_bundle = pickle.load(f)
scaler_A   = scaler_bundle['scaler']
scale_cols = scaler_bundle['scale_cols']

# ── Load full dataset ─────────────────────────────
df = pd.read_csv(data_dir / 'featured_movies.csv')

# ── Recreate same feature set as Step 3 ──────────
# Load preprocessed train to know exact columns
X_train_A = pd.read_csv(data_dir / 'X_train_A.csv', index_col=0)
X_test_A  = pd.read_csv(data_dir / 'X_test_A.csv',  index_col=0)
y_train_reg = pd.read_csv(data_dir / 'y_train_reg.csv', index_col=0).squeeze()
y_test_reg  = pd.read_csv(data_dir / 'y_test_reg.csv',  index_col=0).squeeze()
y_train_cls = pd.read_csv(data_dir / 'y_train_cls.csv', index_col=0).squeeze()
y_test_cls  = pd.read_csv(data_dir / 'y_test_cls.csv',  index_col=0).squeeze()

features_A = X_train_A.columns.tolist()

# ── Build X_full: same columns, same scaler ───────
# Select only feature columns that exist in df
cols_available = [c for c in features_A if c in df.columns]
cols_missing   = [c for c in features_A if c not in df.columns]
if cols_missing:
    print(f'Missing cols (will fill 0): {cols_missing}')

X_full = df[cols_available].copy()
for c in cols_missing:
    X_full[c] = 0
X_full = X_full[features_A]   # enforce column order

# ── Impute nulls ──────────────────────────────────
for col in X_full.columns:
    if X_full[col].isnull().any():
        X_full[col] = X_full[col].fillna(X_full[col].median())

# ── Scale ─────────────────────────────────────────
X_full_scaled = X_full.copy()
sc_present = [c for c in scale_cols if c in X_full_scaled.columns]
X_full_scaled[sc_present] = scaler_A.transform(X_full_scaled[sc_present])

# ── Predict ───────────────────────────────────────
df['predicted_log_revenue'] = lgbm_tuned.predict(X_full_scaled)
df['predicted_revenue']     = np.expm1(df['predicted_log_revenue'])
df['predicted_is_hit']      = xgb_tuned.predict(X_full_scaled)
df['hit_probability']       = xgb_tuned.predict_proba(X_full_scaled)[:, 1]

print(f'df shape with predictions: {df.shape}')
print(f'features_A count: {len(features_A)}')
print(f'Predictions added: predicted_log_revenue, predicted_revenue, predicted_is_hit, hit_probability')
print(f'\nSample predictions:')
print(df[['title','predicted_revenue','predicted_is_hit','hit_probability']].head(5).to_string(index=False))
```

## Step 6.2 — RQ#1: The Success Formula — Top 5 Most Important Features (SHAP)

SHAP (SHapley Additive exPlanations) measures each feature's **individual contribution**.
- **Beeswarm**: feature value (color) and SHAP effect (x-axis)
- **Bar plot**: mean absolute SHAP — overall importance ranking
- Pre-release feature importance is computed separately for Track B

```python
import shap
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from pathlib import Path

# ── SHAP on tuned LightGBM (Track A) ─────────────
explainer = shap.TreeExplainer(lgbm_tuned)
shap_values = explainer(X_test_A)

print(f'SHAP values shape: {shap_values.values.shape}')

# ── Beeswarm: top 20 ──────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 7))
plt.sca(axes[0])
shap.plots.beeswarm(shap_values, max_display=20, show=False)
axes[0].set_title('SHAP Beeswarm — LightGBM Tuned (Track A)\nTop 20 Features', fontsize=11)

# ── Bar plot: mean |SHAP| top 10 ──────────────────
plt.sca(axes[1])
shap.plots.bar(shap_values, max_display=10, show=False)
axes[1].set_title('Mean |SHAP| — Top 10 Features', fontsize=11)

plt.tight_layout()
Path('../reports').mkdir(exist_ok=True)
plt.savefig('../reports/shap_summary.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Top 5 features overall ────────────────────────
mean_shap = pd.Series(
    np.abs(shap_values.values).mean(axis=0),
    index=X_test_A.columns
).sort_values(ascending=False)

print('\nRQ#1 — Top 5 features driving hits (Track A):')
for i, (feat, val) in enumerate(mean_shap.head(5).items(), 1):
    post = ' [post-release]' if feat in ['vote_count','log_vote_count','popularity','vote_average','vote_reliability'] else ''
    print(f'  {i}. {feat:<35} mean|SHAP|={val:.4f}{post}')

# ── Track B: pre-release features only ───────────
X_train_B = pd.read_csv('../data/X_train_B.csv', index_col=0)
X_test_B  = pd.read_csv('../data/X_test_B.csv',  index_col=0)
y_train_reg2 = pd.read_csv('../data/y_train_reg.csv', index_col=0).squeeze()
y_test_reg2  = pd.read_csv('../data/y_test_reg.csv',  index_col=0).squeeze()

import lightgbm as lgb
import warnings
cb_es  = lgb.early_stopping(50, verbose=False)
cb_log = lgb.log_evaluation(period=-1)
with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    lgbm_B = lgb.LGBMRegressor(
        n_estimators=500, learning_rate=0.05, num_leaves=63,
        subsample=0.8, colsample_bytree=0.8,
        random_state=42, n_jobs=-1, verbosity=-1)
    lgbm_B.fit(X_train_B, y_train_reg2,
               eval_set=[(X_test_B, y_test_reg2)],
               callbacks=[cb_es, cb_log])

explainer_B  = shap.TreeExplainer(lgbm_B)
shap_vals_B  = explainer_B(X_test_B)
mean_shap_B  = pd.Series(
    np.abs(shap_vals_B.values).mean(axis=0),
    index=X_test_B.columns
).sort_values(ascending=False)

print('\nRQ#1 — Top 5 pre-release features (Track B):')
for i, (feat, val) in enumerate(mean_shap_B.head(5).items(), 1):
    print(f'  {i}. {feat:<35} mean|SHAP|={val:.4f}')
```

## Step 6.5 — RQ#4 & RQ#12: Flop Analysis and Risk Score

A rule-based **risk score (0–100)** is built from the EDA findings:
- Higher score = higher flop risk
- Validated against actual flops using ROC-AUC

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, roc_curve

# ── Drop NaN targets ──────────────────────────────
df_flop = df.dropna(subset=['is_flop']).copy()

# ── Build risk score ──────────────────────────────
def risk_score(row):
    score = 0.0
    score += 0.30 * row.get('is_indie', 0)
    score += 0.25 * max(
        row.get('genre_Drama', 0),
        row.get('genre_History', 0),
        row.get('genre_Western', 0)
    )
    score += 0.20 * row.get('is_dump_month', 0)
    score += 0.15 * row.get('is_mid_budget', 0)
    score += 0.10 * row.get('is_unknown_director', 0)
    return score * 100

df_flop['risk_score'] = df_flop.apply(risk_score, axis=1)
df['risk_score'] = df.apply(risk_score, axis=1)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('RQ#4 & RQ#12 — Flop Risk Score Analysis', fontsize=13, fontweight='bold')

# ── Risk score distribution ───────────────────────
ax = axes[0]
ax.hist(df_flop.loc[df_flop['is_flop']==0, 'risk_score'], bins=20,
        alpha=0.6, color='#42A5F5', label='Non-flop', density=True)
ax.hist(df_flop.loc[df_flop['is_flop']==1, 'risk_score'], bins=20,
        alpha=0.6, color='#E53935', label='Flop',     density=True)
ax.set_xlabel('Risk Score (0–100)')
ax.set_ylabel('Density')
ax.set_title('Risk Score Distribution')
ax.legend()

# ── ROC curve ─────────────────────────────────────
ax2 = axes[1]
y_true_flop = df_flop['is_flop'].astype(int)
y_score_flop = df_flop['risk_score']
fpr, tpr, _ = roc_curve(y_true_flop, y_score_flop)
auc_risk = roc_auc_score(y_true_flop, y_score_flop)
ax2.plot(fpr, tpr, color='#E53935', lw=2, label=f'AUC={auc_risk:.3f}')
ax2.plot([0,1],[0,1],'k--',lw=1)
ax2.set_xlabel('FPR'); ax2.set_ylabel('TPR')
ax2.set_title('ROC — Risk Score vs is_flop')
ax2.legend()

# ── Risk score by flop status (box) ───────────────
ax3 = axes[2]
bp = ax3.boxplot(
    [df_flop.loc[df_flop['is_flop']==0, 'risk_score'],
     df_flop.loc[df_flop['is_flop']==1, 'risk_score']],
    labels=['Non-flop','Flop'], patch_artist=True, showfliers=False
)
bp['boxes'][0].set_facecolor('#90CAF9')
bp['boxes'][1].set_facecolor('#EF9A9A')
ax3.set_ylabel('Risk Score')
ax3.set_title('Risk Score by Outcome')

plt.tight_layout()
plt.savefig('../reports/risk_score.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Top 10 highest risk films ─────────────────────
top_risk = df_flop.nlargest(10, 'risk_score')[[
    'title','risk_score','is_flop','revenue','predicted_revenue'
]].copy()
top_risk['revenue']           = (top_risk['revenue'] / 1e6).round(1)
top_risk['predicted_revenue'] = (top_risk['predicted_revenue'] / 1e6).round(1)
top_risk['outcome'] = top_risk['is_flop'].map({1:'FLOP',0:'non-flop'})

print(f'Risk score AUC (predicting flop): {auc_risk:.4f}\n')
print('Top 10 highest risk films:')
print(f'{"Title":<40} {"Risk":>5} {"Outcome":<10} {"Actual$M":>9} {"Pred$M":>7}')
print('-'*75)
for _, r in top_risk.iterrows():
    print(f'{r["title"]:<40} {r["risk_score"]:>5.0f} {r["outcome"]:<10} '
          f'{r["revenue"]:>9.1f} {r["predicted_revenue"]:>7.1f}')

print(f'\nFlop rate by risk tier:')
df_flop['risk_tier'] = pd.cut(df_flop['risk_score'], bins=[0,25,50,75,100],
                               labels=['Low(0-25)','Mid(25-50)','High(50-75)','VHigh(75+)'])
print(df_flop.groupby('risk_tier', observed=True)['is_flop']
            .agg(['mean','count'])
            .rename(columns={'mean':'flop_rate','count':'n_films'}).to_string())
```

### 📋 Step 6.5 — RQ#4 & RQ#12 Answer: Flop Risk

|-----------|-----------|-------------|
| Low (0-25) | 24.1% | 1323 |
| Mid (25-50) | 31.8% | 614 |
| High (50-75) | **46.3%** | 123 |

The rule-based risk score gets AUC=0.59 — weaker than the ML model but interpretable.
The flop rate in the High tier is 2× that of the Low tier.

> **Riskiest combination:** independent production + Drama/Western genre + September/January release + mid budget + unknown director.

## Step 6.6 — RQ#5: Optimal Runtime

- **Scatter plot**: actual vs predicted revenue

```python
import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, r2_score
from pathlib import Path

data_dir = Path('../data')
model_dir = Path('../models')

X_test = pd.read_csv(data_dir / 'X_test_A.csv', index_col=0)
y_test = pd.read_csv(data_dir / 'y_test_reg.csv', index_col=0).squeeze()

with open(model_dir / 'lgbm_tuned.pkl', 'rb') as f:
    lgbm = pickle.load(f)

with open(model_dir / 'scaler_A.pkl', 'rb') as f:
    scaler_obj = pickle.load(f)
scaler = scaler_obj['scaler']
scale_cols = scaler_obj['scale_cols']

X_test_s = X_test.copy()
X_test_s[scale_cols] = scaler.transform(X_test[scale_cols])

y_pred_log = lgbm.predict(X_test_s)

# Back-transform to dollars
y_true_usd = np.expm1(y_test.values)
y_pred_usd = np.expm1(y_pred_log)

mae_usd = mean_absolute_error(y_true_usd, y_pred_usd)
r2 = r2_score(y_true_usd, y_pred_usd)

print(f'MAE  : ${mae_usd/1e6:.1f}M')
print(f'R²   : {r2:.4f}')

# Scatter plot
fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(y_true_usd / 1e6, y_pred_usd / 1e6, alpha=0.3, s=18, color='steelblue')
lim = max(y_true_usd.max(), y_pred_usd.max()) / 1e6
ax.plot([0, lim], [0, lim], 'r--', lw=1.5, label='Perfect fit')
ax.set_xlabel('Actual Revenue ($M)')
ax.set_ylabel('Predicted Revenue ($M)')
ax.set_title(f'LightGBM Tuned — Actual vs Predicted\nMAE=${mae_usd/1e6:.1f}M  R²={r2:.3f}')
ax.legend()
plt.tight_layout()
Path('../reports').mkdir(exist_ok=True)
plt.savefig('../reports/eval_scatter.png', dpi=150)
plt.show()
```

## Step 7.1 — Track B Data: Pre-Release Features

In this step we load the Track B dataset, which only contains features **knowable before release**.
We verify that the columns causing data leakage — `vote_average`, `popularity`, `vote_count`, `log_vote_count`, `vote_reliability` — are **absent**.

```python
import numpy as np
import pandas as pd
import pickle
from pathlib import Path

data_dir  = Path('../data')
model_dir = Path('../models')

X_train_B = pd.read_csv(data_dir / 'X_train_B.csv', index_col=0)
X_test_B  = pd.read_csv(data_dir / 'X_test_B.csv',  index_col=0)
y_train_reg = pd.read_csv(data_dir / 'y_train_reg.csv', index_col=0).squeeze()
y_test_reg  = pd.read_csv(data_dir / 'y_test_reg.csv',  index_col=0).squeeze()
y_train_cls = pd.read_csv(data_dir / 'y_train_cls.csv', index_col=0).squeeze()
y_test_cls  = pd.read_csv(data_dir / 'y_test_cls.csv',  index_col=0).squeeze()

with open(model_dir / 'scaler_B.pkl', 'rb') as f:
    scaler_obj_B = pickle.load(f)
scaler_B     = scaler_obj_B['scaler']
scale_cols_B = scaler_obj_B['scale_cols']

# Apply scaler
X_train_Bs = X_train_B.copy()
X_test_Bs  = X_test_B.copy()
X_train_Bs[scale_cols_B] = scaler_B.transform(X_train_B[scale_cols_B])
X_test_Bs[scale_cols_B]  = scaler_B.transform(X_test_B[scale_cols_B])

# Assert no leakage
leakage_cols = ['vote_average', 'popularity', 'vote_count', 'log_vote_count', 'vote_reliability']
found = [c for c in leakage_cols if c in X_train_B.columns]
assert len(found) == 0, f'Leakage columns found: {found}'
print('Leakage check: PASSED')
print(f'X_train_B: {X_train_B.shape}  |  X_test_B: {X_test_B.shape}')
print(f'Features ({len(X_train_B.columns)}):', list(X_train_B.columns))
```

## Step 7.2 — LightGBM Regressor (Track B Baseline)

A LightGBM regression model is trained using only pre-release features.
Target: `log_revenue` — MAE is computed in dollars after back-transforming.

```python
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error, r2_score

lgbm_B = lgb.LGBMRegressor(
    n_estimators=500, learning_rate=0.05, num_leaves=63,
    min_child_samples=20, subsample=0.8, colsample_bytree=0.8,
    random_state=42, n_jobs=-1, verbose=-1
)
lgbm_B.fit(
    X_train_Bs, y_train_reg,
    eval_set=[(X_test_Bs, y_test_reg)],
    callbacks=[lgb.early_stopping(50, verbose=False), lgb.log_evaluation(100)]
)

y_pred_log = lgbm_B.predict(X_test_Bs)
rmse_log   = np.sqrt(np.mean((y_test_reg.values - y_pred_log)**2))
mae_usd    = mean_absolute_error(np.expm1(y_test_reg), np.expm1(y_pred_log))
r2         = r2_score(np.expm1(y_test_reg), np.expm1(y_pred_log))

print(f'LightGBM Track B Baseline')
print(f'  RMSE (log) : {rmse_log:.4f}')
print(f'  MAE        : ${mae_usd/1e6:.1f}M')
print(f'  R²         : {r2:.4f}')
```

```python
import xgboost as xgb
from sklearn.metrics import roc_auc_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

xgb_B = xgb.XGBClassifier(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    eval_metric='auc', early_stopping_rounds=50,
    random_state=42, n_jobs=-1, verbosity=0
)
xgb_B.fit(
    X_train_Bs, y_train_cls,
    eval_set=[(X_test_Bs, y_test_cls)],
    verbose=100
)

y_prob = xgb_B.predict_proba(X_test_Bs)[:, 1]
y_pred_cls = (y_prob >= 0.5).astype(int)
auc = roc_auc_score(y_test_cls, y_prob)
f1  = f1_score(y_test_cls, y_pred_cls)

print(f'XGBoost Track B Baseline')
print(f'  AUC : {auc:.4f}')
print(f'  F1  : {f1:.4f}')

fig, ax = plt.subplots(figsize=(4, 3))
cm = confusion_matrix(y_test_cls, y_pred_cls)
ConfusionMatrixDisplay(cm, display_labels=['Non-hit', 'Hit']).plot(ax=ax, colorbar=False)
ax.set_title('XGBoost Track B — Confusion Matrix')
plt.tight_layout()
plt.show()
```

20 iterations with RandomizedSearchCV, 5-fold CV. Target: minimize `neg_root_mean_squared_error`.

Track A models (which include post-release features) are compared against the zero-leakage Track B models.

```python
# Load Track A tuned models for comparison
with open(model_dir / 'lgbm_tuned.pkl', 'rb') as f:
    lgbm_A_tuned = pickle.load(f)
with open(model_dir / 'xgb_cls_tuned.pkl', 'rb') as f:
    xgb_A_tuned = pickle.load(f)
with open(model_dir / 'scaler_A.pkl', 'rb') as f:
    scaler_obj_A = pickle.load(f)

X_test_A  = pd.read_csv(data_dir / 'X_test_A.csv',  index_col=0)
X_test_As = X_test_A.copy()
sc_A = scaler_obj_A['scaler']
sc_A_cols = scaler_obj_A['scale_cols']
X_test_As[sc_A_cols] = sc_A.transform(X_test_A[sc_A_cols])

# Track A regression
yp_A_log = lgbm_A_tuned.predict(X_test_As)
mae_A  = mean_absolute_error(np.expm1(y_test_reg), np.expm1(yp_A_log))
r2_A   = r2_score(np.expm1(y_test_reg), np.expm1(yp_A_log))

# Track A classification
yp_A_prob = xgb_A_tuned.predict_proba(X_test_As)[:, 1]
auc_A = roc_auc_score(y_test_cls, yp_A_prob)

# Collect results
results = [
    ('LightGBM  Track A (tuned)', f'${mae_A/1e6:.1f}M',  f'{r2_A:.3f}',  '—'),
    ('LightGBM  Track B (tuned)', f'${mae_t/1e6:.1f}M',  f'{r2_t:.3f}',  '—'),
    ('XGBoost   Track A (tuned)', '—',                    '—',            f'{auc_A:.4f}'),
    ('XGBoost   Track B (tuned)', '—',                    '—',            f'{auc_t:.4f}'),
]
cmp = pd.DataFrame(results, columns=['Model', 'MAE ($)', 'R²', 'AUC'])
print(cmp.to_string(index=False))

# Delta display
print(f'\nMAE delta  (B - A): +${(mae_t - mae_A)/1e6:.1f}M  (cost of removing post-release features)')
print(f'AUC delta  (B - A): {auc_t - auc_A:+.4f}')
```

### 📋 Step 7.6 — Summary

**The Track A vs Track B gap** = the value of post-release features.
The Track B model is an honest pre-release predictor: it represents the maximum performance achievable in real life before a movie is released.
MAE gap ~$15–25M, AUC gap ~0.05–0.08 expected.
